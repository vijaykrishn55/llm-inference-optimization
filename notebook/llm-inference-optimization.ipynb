```python
%%bash
cd /kaggle/working/llm-inference-optimization

./.venv-vllm/bin/python - <<'PY'
from vllm import LLM, SamplingParams

model_path = "artifacts/quantized_model"

print("Loading quantized model into vLLM...")

llm = LLM(
    model=model_path,
    tensor_parallel_size=1,
    dtype="half",
    max_model_len=1024,
    gpu_memory_utilization=0.70,
    max_num_seqs=4,
)

print("Model loaded successfully.")

params = SamplingParams(temperature=0.0, max_tokens=64)
outputs = llm.generate(
    ["Explain simply why model quantization reduces memory usage."],
    params,
)

for output in outputs:
    print("\nGenerated text:")
    print(output.outputs[0].text)
    print("Prompt tokens:", len(output.prompt_token_ids))
    print("Generated tokens:", len(output.outputs[0].token_ids))

print("\nQUANTIZED INFERENCE TEST PASSED")
PY
```

```python
import json
import urllib.request
import urllib.error

BASE_URL = "http://127.0.0.1:8000"

def post_json(url, payload):
    request = urllib.request.Request(
        url,
        data=json.dumps(payload).encode("utf-8"),
        headers={"Content-Type": "application/json"},
    )
    with urllib.request.urlopen(request, timeout=300) as response:
        return json.loads(response.read().decode("utf-8"))

# Get the model ID exposed by vLLM
models = json.load(
    urllib.request.urlopen(f"{BASE_URL}/v1/models", timeout=10)
)
model_id = models["data"][0]["id"]

# Generate a sufficiently long response to collect metrics
result = post_json(
    f"{BASE_URL}/v1/chat/completions",
    {
        "model": model_id,
        "messages": [
            {
                "role": "user",
                "content": (
                    "Explain how a transformer language model generates text, "
                    "covering tokenization, attention, next-token prediction, "
                    "and autoregressive decoding. Be detailed."
                ),
            }
        ],
        "max_tokens": 128,
        "temperature": 0,
    },
)

print("MODEL:", model_id)
print("\nRESPONSE:")
print(result["choices"][0]["message"]["content"])

usage = result.get("usage", {})
print("\nTOKEN USAGE:", usage)

metrics = result.get("metrics") or {}
spec = metrics.get("speculative_decoding")

print("\nSPECULATIVE DECODING METRICS:")
if spec:
    print(json.dumps(spec, indent=2))
    print(
        f"\nDraft acceptance rate: "
        f"{spec['draft_acceptance_rate']:.2%}"
    )
    print(
        f"Mean acceptance length: "
        f"{spec['mean_acceptance_length']:.2f} tokens per verification step"
    )
else:
    print("Metrics missing. Check the server log and response format.")
```
